```python
import torch
import h5py
```

```python
dataset = 'data_Synthetic_Lorenz_Offmanifold_Unidir_Kicks'
# dataset = 'data_Synthetic_SimplePendulum_OffManifold_Perturbation'

train_set = torch.load(f"./{dataset}/data_train_20ms.pt")
valid_set = torch.load(f"./{dataset}/data_valid_20ms.pt")
test_set = torch.load(f"./{dataset}/data_test_adaptation_20ms.pt")
```

```python
filename = f"../../datasets/{dataset}.h5"

with h5py.File(filename, 'w') as h5file:
    # variables needed for training
    h5file.create_dataset('train_encod_data', data=train_set['y_obs'])
    h5file.create_dataset('valid_encod_data', data=valid_set['y_obs'])
    h5file.create_dataset('train_recon_data', data=train_set['y_obs'])
    h5file.create_dataset('valid_recon_data', data=valid_set['y_obs'])
    h5file.create_dataset('train_behavior', data=train_set['velocity'])
    h5file.create_dataset('valid_behavior', data=valid_set['velocity'])
    # variables needed for post analysis
    h5file.create_dataset('train_epoch', data=train_set['is_perturbed'])
    h5file.create_dataset('valid_epoch', data=valid_set['is_perturbed'])
```

```python
filename = f"../../datasets/{dataset}_test.h5"

with h5py.File(filename, 'w') as h5file:
    # variables needed for training
    h5file.create_dataset('train_encod_data', data=train_set['y_obs'])
    h5file.create_dataset('valid_encod_data', data=test_set['y_obs'])
    h5file.create_dataset('train_recon_data', data=train_set['y_obs'])
    h5file.create_dataset('valid_recon_data', data=test_set['y_obs'])
    h5file.create_dataset('train_behavior', data=train_set['velocity'])
    h5file.create_dataset('valid_behavior', data=test_set['velocity'])
    # variables needed for post analysis
    h5file.create_dataset('train_epoch', data=train_set['is_perturbed'])
    h5file.create_dataset('valid_epoch', data=test_set['is_perturbed'])
```
